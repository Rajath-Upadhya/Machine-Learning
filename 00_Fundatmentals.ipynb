Shape errors

```python
# Shapes for matrix multiplication
tensor_A = torch.tensor([[1, 2],
                         [3, 4],
                         [5, 6]])

tensor_B = torch.tensor([[7, 10],
                         [8, 11],
                         [9, 12]])

torch.mm(tensor_A, tensor_B) # torch.mm is the same as torch.matmul

# Error: Inner dimension do not match: (3,2) * (3,2)
# It needs to be: (3,2) * (2,3) leading to 3*3 matrix (Outer dimension)
```

```python
# To fix, use transpose: .T

torch.mm(tensor_A, tensor_B.T)
```

```python
tensor_B.T.shape
```

```python
# The matrix multiplication
print(f"Original shapes: tensor_A = {tensor_A.shape}, tensor_B = {tensor_B.shape}")
print(f"New shapes: tensor_A = {tensor_A.shape}, tensor_B.T = {tensor_B.T.shape}")
print(f"Multiplying: {tensor_A.shape} * {tensor_B.T.shape} <- inner dimensions must match")
print("Output:\n")
print(torch.mm(tensor_A, tensor_B.T))
print(f"Output shape: {torch.mm(tensor_A, tensor_B.T).shape}")
```

## Reshaping, stacking, squeezing and unsqueezing tensors

* Reshaping - reshapes an input tensor to a defined shape
* View - Return a view of an input tensor of certain shape but keep the same memory as the original tensor
* Stacking - combine multuple tensors on top of each other (vstack), or side by side (hstack)
* Squeeze - removes all `1` dimensions from a tensor
* Unsqueeze - add a `1` dimension to a target tensor
* Permute - Return a view of the input with dimensions permuted (swapper) in a certain way


```python
y = torch.arange(1, 10)
y, y.shape
```

```python
# Add an extra dimension
y_reshaped = y.reshape(1, 9)
y_reshaped, y_reshaped.shape
```

```python
y_reshaped = y.reshape(9, 1)
y_reshaped, y_reshaped.shape
```

```python
y_reshaped = y.reshape(3, 3)
y_reshaped, y_reshaped.shape
```

```python
# Change the view
z = y.view(1,9)
z, z.shape
```

### Changing z changes y because a view of a tensor shares the same memory as the original y

```python
z[:,0] = 5
z, y
```

```python
# Stack tensors on top of each other
x_stacked = torch.stack([y, y, y, y], dim=0)
x_stacked
```

```python
x_stacked = torch.stack([y, y, y, y], dim=1)
x_stacked
```

```python
x_vstacked = torch.vstack([y, y, y, y])
x_vstacked, x_vstacked.shape
```

```python
x_hstacked = torch.hstack([y, y, y, y])
x_hstacked, x_hstacked.shape
```

```python
s = y_reshaped.reshape(1,9)
s, s.shape
```

```python
# torch.squeeze() - removes all single dimensions from a target tensor
print(f"Previous tensor: {s}")
print(f"Previous shape: {s.shape}")

# Remove extra dimensions from s
y_squeezed = s.squeeze()
print(f"\nNew tensor: {y_squeezed}")
print(f"New shape: {y_squeezed.shape}")
```

```python
# torch.unsqueeze() - adds a single dimension to target tensor at a specific dim
print(f"Previous tensor: {y_squeezed}")
print(f"Previous shape: {y_squeezed.shape}")

# Add an extra dimension with unsqueeze
y_unsqueezed = y_squeezed.unsqueeze(dim=0)
print(f"\nNew tensor: {y_unsqueezed}")
print(f"New shape: {y_unsqueezed.shape}")
```

```python
# torch.permute() - rearranges the dimensions of a target tensor in a specified order
x_original = torch.rand(size=(224,224,3)) # Color channels

# Permite the original rensor to rearrange the axis or dim order
x_permuted = x_original.permute(2, 0, 1) # Shifts axis 0->1, 1->2, 2->0

print(f"Previous shape: {x_original.shape}")
print(f"New shape: {x_permuted.shape}")

x_permuted
```

```python
x_original[0, 0, 0] = 7284
x_permuted = x_original.permute(2, 0, 1)
x_original[0,0,0], x_permuted[0, 0, 0]
```

# Pytorch tensors and Numpy

* Data in numpy array to Pytorch tensor -> `torch.from_numpy(ndarray)`
* Pytorch tenor to numpy -> `torch.Tensor.numpy()`

```python
# NumPy array to tensor
import torch
import numpy as np

array = np.arange(1.0, 8.0)
tensor = torch.from_numpy(array) # warning: when converting from numpy -> pytorch, pytorch reflects numpy's default datatype of float64 unless specified otherwise
array, tensor
```

```python
# From float64 to float32
tensor_float32 = torch.from_numpy(array).type(torch.float32)
tensor_float32.dtype
```

```python
# If we change the array, it does not change the tensor created from it.
# Example if we add 1 to array, the created tensor remains same.
# array and tensor do not share same memeory

array = array + 1
array, tensor
```

```python
# Tensor to NumPy
tensor = torch.ones(7)
numpy_tensor = tensor.numpy()
tensor, numpy_tensor
# DataType is carried from torch to NumPy
```

```python
# Change the tensor, and check numpy_tensor
tensor = tensor + 1
tensor, numpy_tensor
```

## Reproducibility

- Taking random out of random

- Neural network learns: start with random numbers -> tensor operations -> update random numbers to try and make them better representation of the data -> repeat

* To reduce the randomness in neural networks and PyTorch, **random seed** concept used.

* https://docs.pytorch.org/docs/2.14/notes/randomness.html


```python
import torch

# Create two random tensors
random_tensor_A = torch.rand(3,4)
random_tensor_B = torch.rand(3,4)

print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B)
```

```python
# Random but reproducible tensors
import torch

# Set the random seed
Random_seed = 10
torch.manual_seed(Random_seed)
random_tensor_C = torch.rand(3,4)

# If not set again, it won't have random_seed to tensor_D
torch.manual_seed(Random_seed)
random_tensor_D = torch.rand(3,4)

print(random_tensor_C)
print(random_tensor_D)
print(random_tensor_C == random_tensor_D)
```

## Check for GPU access with PyTorch

```python
# No setup required
import torch
torch.cuda.is_available()
```

```python
# Setup device agnostic code
device = "cuda" if torch.cuda.is_available() else "cpu"
device
```

```python
# Device count
torch.cuda.device_count()
```

```python
# Tensors and models on the GPU for faster computations
tensor = torch.tensor([1, 2, 3])

print(tensor, tensor.device)
```

```python
# Move tensor to GPU if available

tensor_on_gpu = tensor.to(device)
tensor_on_gpu
```

```python
# Move tensor back to the CPU
# If tensor is on GPU, can't transform it to NumPy
tensor_on_gpu.numpy()
```

```python
# To fix the GPU tensor with NumPy issue, we can first set it to the CPU
tensor_back_on_CPU = tensor_on_gpu.cpu().numpy()
tensor_back_on_CPU
```
