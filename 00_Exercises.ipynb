Create two random tensors of shape (2, 3) and send them both to the GPU

```python
Random_tensor_5 = torch.rand(2,3).to(device)
Random_tensor_6 = torch.rand(2,3).to(device)
Random_tensor_6
```

Perform a matrix multiplication on the tensors

```python
# Multiplication in GPU
multiplication_on_gpu = torch.mm(Random_tensor_5, Random_tensor_6.T)
multiplication_on_gpu
```

Find the maximum and minimum values of the output

```python
# Finding maximum and minimum values
torch.max(multiplication_on_gpu), torch.min(multiplication_on_gpu)
```

Find the maximum and minimum index values of the output

```python
multiplication_on_gpu.argmax(), multiplication_on_gpu.argmin()
```

Make a random tensor with shape `(1, 1, 1, 10)` and then create a new tensor with all the 1 dimensions removed to be left with a tensor of shape `(10)`. Set the seed to `7` when you create it and print out the first tensor and it's shape as well as the second tensor and it's shape.

```python
import torch
random_seed = 7
torch.manual_seed(random_seed)
random_tensor_1 = torch.rand(1,1,1,10)
print(f"Tensor: {random_tensor_1}")
print(f"Shape: {random_tensor_1.shape}")
print(f"Squeezed tensor: {random_tensor_1.squeeze()}")
print(f"Squeezed shape: {random_tensor_1.squeeze().shape}")
```
